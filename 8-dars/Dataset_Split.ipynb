```python

import torch
from torchvision import datasets, transforms
from torchvision.datasets import MNIST
```

```python
from torchvision import transforms as T
mean = [0.5]
std = [0.5]
tfs = T.Compose([T.ToTensor(), T.Normalize(mean = mean, std = std) ])
train_dataset = MNIST(root="datasets", train=True, download=True, transform=tfs)
test_dataset = MNIST(root="datasets", train=False, download=True, transform=tfs)
```

```python
from torch.utils.data import random_split

len(test_dataset), len(train_dataset)

# 1. MNIST datasetni yuklab olish
transform = transforms.ToTensor()

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

print("Umumiy dataset:", len(train_dataset))

# 2. Split variantlari
split_sizes = [
    (40000, 20000),
    (50000, 10000),
    (30000, 30000),
    (55000, 5000),
    (58000, 2000)
]

# 3. Har bir variantni random_split yordamida ajratish
for i, (train_size, val_size) in enumerate(split_sizes, start=1):

    generator = torch.Generator().manual_seed(1)

    train_set, val_set = random_split(
        train_dataset,
        [train_size, val_size],
        generator=generator
    )

    print(f"\nVariant {i}:")
    print("Train Dataset:", len(train_set))
    print("Validation Dataset:", len(val_set))
```

#  Hidden Layers Sonini O’zgartirish:

```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_hidden, output_size): # asoiy uzgarishlar
        super().__init__()

        self.input_layer = nn.Linear(input_size, hidden_size)

        self.hidden_layers = nn.ModuleList([ # hidden layerni uzgartirish
            nn.Linear(hidden_size, hidden_size)
            for _ in range(num_hidden)
        ])

        self.output_layer = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = torch.relu(self.input_layer(x))

        for layer in self.hidden_layers:
            x = torch.relu(layer(x))

        x = self.output_layer(x)

        return x
```

```python
# num_hidden yordamida 2, 4 va 6 ta hidden layerli modellarni yaratib, ularni solishtirish kerak.

model_2 = MyModel(
    input_size=784,
    hidden_size=128,
    num_hidden=2,
    output_size=10
)

model_4 = MyModel(
    input_size=784,
    hidden_size=128,
    num_hidden=4,
    output_size=10
)

model_6 = MyModel(
    input_size=784,
    hidden_size=128,
    num_hidden=6,
    output_size=10
)
```

```python
# Modellarni kuramiz.
print("2 hidden layers:")
print(model_2)

print("\n4 hidden layers:")
print(model_4)

print("\n6 hidden layers:")
print(model_6)
```

```python
# Parametrlar sonini uzgarishini kuramiz.
def count_parameters(model):
    return sum(p.numel() for p in model.parameters())

print("2 hidden:", count_parameters(model_2))
print("4 hidden:", count_parameters(model_4))
print("6 hidden:", count_parameters(model_6))
```
