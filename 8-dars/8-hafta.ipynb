```python
from torch.nn import *

# Moduledan inheritance qilganda 2 ta funciton o'zgartirilishi kerak:
# init
# forward - classni chaqirganda qanday amallar bajarishini ko'rsatadi.


import torch
from torch.nn import *

# Model yaratish uchun nn.Module dan inheritance qilamiz
class CustomModel(Module):
    def __init__(self, in_fs, out_fs, klasslar_soni):
        super().__init__()

        self.linear_1 = Linear(in_features=in_fs, out_features=out_fs)
        self.linear_2 = Linear(in_features=out_fs, out_features=out_fs * 4)
        self.linear_3 = Linear(in_features=out_fs * 4, out_features=klasslar_soni)

        self.relu = ReLU()

    def forward(self, x):
        bs = x.shape[0]

        # 4D tensor -> 2D tensor
        x = x.view(bs, -1)

        out = self.linear_1(x)
        out = self.relu(out)

        out = self.linear_2(out)
        out = self.relu(out)

        out = self.linear_3(out)

        return out


inp = torch.rand(1, 1, 28, 28)

model = CustomModel(
    in_fs=784,
    out_fs=200,
    klasslar_soni=10
)

result = model(inp)

print(result)
print(result.shape)
```
