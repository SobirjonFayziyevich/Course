Learning rate katta bo‘lsa, parametrlar juda katta qadam bilan yangilanadi. Natijada model optimal nuqtadan oshib ketishi va loss ortib ketishi mumkin.
Learning rate kichik bo‘lsa, parametrlar sekin yangilanishini kurdim. Model barqarorroq o‘rganadi, ammo optimal natijaga yetishi uchun ko‘proq iteratsiya talab etilarkan.
