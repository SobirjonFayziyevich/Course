```python
import torch
import torchvision
from torch.nn import Module, Linear, ReLU, LeakyReLU, ELU, PReLU, GELU, SiLU, Softmax, Sigmoid
```

# ReLu Hidden Layer

```python
## 1 - model 2 hidden layer ReLU
class CustomModel_1(Module):
    def __init__(self,in_fs, out_fs, n_cls):
        super().__init__()
        self.hidden1 = Linear(in_features=in_fs, out_features=out_fs)
        self.activation1 = ReLU()

        self.hidden2 = Linear(in_features=out_fs, out_features=out_fs)
        self.activation2 = ReLU()
        self.classifier = Linear(in_features=out_fs, out_features=n_cls)

    def forward(self, input):
        bs = input.shape[0]
        input = input.view(bs, -1) # 2D ga o'zgartirish uchun.

        out = self.hidden1(input)
        out = self.activation1(out)    
        out = self.hidden2(out)
        out = self.activation2(out)
        out = self.classifier(out)
        return out
input = torch.rand(1, 1, 28, 28)
out_function = 200
model = CustomModel_1(in_fs = (input.shape[1] * input.shape[2] * input.shape[3]), out_fs = out_function, n_cls = 10)
model(input)
```

```python
### 4 hidden layer - ReLu out with Softmax
class CustomModel_4H_Soft(Module):
    def __init__(self, in_fs, out_fs, n_cls):
        super().__init__()
        self.hidden1 = Linear(in_features=in_fs, out_features=out_fs)
        self.act1 = ReLU()

        self.hidden2 = Linear(in_features=out_fs, out_features=out_fs)
        self.act2 = ReLU()

        self.hidden3 = Linear(in_features=out_fs, out_features=out_fs)
        self.act3 = ReLU()

        self.hidden4 = Linear(in_features=out_fs, out_features=out_fs)
        self.act4 = ReLU()

        self.classifier = Linear(in_features=out_fs, out_features=n_cls)
        self.act5 = Softmax(dim=1)

    def forward(self, inp):
        bs = inp.shape[0]
        inp = inp.view(bs, -1)

        out = self.hidden1(inp)
        out = self.act1(out)
        out = self.hidden2(out)
        out = self.act2(out)
        out = self.hidden3(out)
        out = self.act3(out)
        out = self.hidden4(out)
        out = self.act4(out)
        out = self.classifier(out)
        out = self.act5(out)
        return out

inp = torch.rand(1,1,28,28)
out_fs = 200
m = CustomModel_4H_Soft(in_fs = (inp.shape[1] * inp.shape[2] * inp.shape[3]), out_fs = out_fs, n_cls = 10)
pred = m(inp)
pred
```

```python
### 4 hidden layer - ReLu out with Sigmoid
class CustomModel_4H_Sig(Module):
    def __init__(self, in_fs, out_fs, n_cls):
        super().__init__()
        self.hidden1 = Linear(in_features=in_fs, out_features=out_fs)
        self.act1 = ReLU()

        self.hidden2 = Linear(in_features=out_fs, out_features=out_fs)
        self.act2 = ReLU()

        self.hidden3 = Linear(in_features=out_fs, out_features=out_fs)
        self.act3 = ReLU()

        self.hidden4 = Linear(in_features=out_fs, out_features=out_fs)
        self.act4 = ReLU()
        self.classifier = Linear(in_features=out_fs, out_features=n_cls)
        self.act5 = Sigmoid()

    def forward(self, inp):
        bs = inp.shape[0]
        inp = inp.view(bs, -1)

        out = self.hidden1(inp)
        out = self.act1(out)
        out = self.hidden2(out)
        out = self.act2(out)
        out = self.hidden3(out)
        out = self.act3(out)
        out = self.hidden4(out)
        out = self.act4(out)
        out = self.classifier(out)
        out = self.act5(out)
        return out

inp = torch.rand(1,1,28,28)
out_fs = 200
m = CustomModel_4H_Sig(in_fs = (inp.shape[1] * inp.shape[2] * inp.shape[3]), out_fs = out_fs, n_cls = 1)
m(inp)
```
