# DataLoader

```python
from torch.utils.data import DataLoader
transform = transforms.ToTensor()
train_dataset = MNIST(root = "datasets", train = True, download=True, transform=transform)
```

```python
batch_size = [10,25,64,128,256,512]
results = {
    "Batch Size": [],
    "Total Batches": []
}
for bs in batch_size:
    loader = DataLoader(train_dataset,batch_size=bs, shuffle=True)
    results["Batch Size"].append(bs)
    results["Total Batches"].append(len(loader))
    
```
